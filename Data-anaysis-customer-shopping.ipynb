```python
# Replace null or missing values with median over mean because mean can be affected by outliers and that can skew the data.
# Whereas median is robust to outliers
# We will find median review rating of each category and replace it with the missing value
df ['Review Rating'] = df['Review Rating'].fillna(df.groupby('Category')['Review Rating'].transform('median'))
```

```python
df.isnull().sum()
```

```python
#Replace the upper case column names to lower case and into snake case(no spaces but underscore)
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# Create a new column age_group
#qcut divides a numeric column into quantile-based bins.
  #q=4 → divides the data into 4 equal-sized groups (quartiles).
  #Each group contains ~25% of the observations.
labels = ['Young Adult', 'Adult','Middle aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create a column purchase_frequency_days
# map a dictionary to the already existing column (frequency_of_purchases)
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly' : 7,
    'Monthly':30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually':365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
# The result True means that both of these columns have exact same info
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
#axis=1 is columns
#axis=0 is rows
df = df.drop('promo_code_used', axis=1)
```
